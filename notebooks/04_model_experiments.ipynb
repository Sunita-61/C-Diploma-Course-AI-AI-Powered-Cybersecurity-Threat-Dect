# 04. Model Experiments & Anomaly Detection
### AI-Powered Cybersecurity Threat Detection System

This notebook demonstrates training Random Forest, Logistic Regression, and Isolation Forest models, and evaluating their detection metrics.

```python
import sys
from pathlib import Path
import pandas as pd

PROJECT_ROOT = Path("..").resolve()
sys.path.insert(0, str(PROJECT_ROOT))

from src.threat_detection.simulation import generate_synthetic_events
from src.threat_detection.preprocessing import clean_dataframe, separate_features_and_target
from src.threat_detection.features import engineer_domain_features
from src.threat_detection.model import train_supervised_model
from src.threat_detection.anomaly_detector import fit_anomaly_detector, score_anomalies

df = generate_synthetic_events(count=300, seed=42)
clean = clean_dataframe(df)
feat = engineer_domain_features(clean)
X, y = separate_features_and_target(feat, target_column='label')

# Train Random Forest
rf_pipe, rf_meta, (X_tr, X_te, y_tr, y_te) = train_supervised_model(X, y, model_type='random_forest')
print(f"Random Forest trained on {len(X_tr)} samples. Test accuracy: {rf_pipe.score(X_te, y_te):.4f}")

# Fit Isolation Forest
iso_pipe, iso_meta = fit_anomaly_detector(X[y == 0], contamination=0.05)
flags, scores = score_anomalies(iso_pipe, X_te)
print(f"Isolation Forest scored {len(flags)} test samples; flagged {flags.sum()} anomalies.")
```
